## 1. Asking yes/no questions — comparisons and booleans

Before Python can *decide* anything, it needs to *ask a question* whose answer is `True` or `False`. Those two special values are called **booleans**, and every decision in every program on Earth ultimately boils down to them.

| Operator | Question it asks   | Example    | Answer  |
|----------|--------------------|------------|---------|
| `==`     | equal?             | `5 == 5`   | `True`  |
| `!=`     | not equal?         | `5 != 3`   | `True`  |
| `<`      | less than?         | `3 < 5`    | `True`  |
| `>`      | greater than?      | `3 > 5`    | `False` |
| `<=`     | less or equal?     | `5 <= 5`   | `True`  |
| `>=`     | greater or equal?  | `4 >= 6`   | `False` |

> ⚠️ **The classic beginner mix-up:** one `=` **assigns** a value (`x = 5` means "store 5 in x"). Two `==` **compares** (`x == 5` asks "is x equal to 5?"). Python will refuse to run `if x = 5:` — a helpful safety net.

You can also **combine** questions with `and`, `or`, `not` — they work just like in English.

````python
age = 25

print(age > 18)        # is age greater than 18?
print(age == 30)       # is age exactly 30?
print(age != 30)       # is age NOT 30?

# Combining questions — read them out loud, they say what they mean:
has_ticket = True
print(age >= 18 and has_ticket)   # both must be True
print(age >= 65 or age <= 12)     # at least one must be True
print(not has_ticket)             # flips True <-> False
````

## 2. `if / elif / else` — teaching Python to decide

### The intuition: a decision tree

Think of how a doctor triages patients, or how you decide what to wear: you walk down a list of questions **in order** and take the **first branch that applies** — then you stop asking.

```text
                 Is it below 5°C? ──── yes ──→  "winter coat"
                        │ no
                 Is it below 15°C? ─── yes ──→  "light jacket"
                        │ no
                        └────────────────────→  "t-shirt"
```

That is *exactly* what `if / elif / else` does: `if` asks the first question, each `elif` ("else-if") asks the next one, and `else` catches everything that fell through.

### Indentation — how Python knows what belongs to a decision

This is the #1 stumbling block for beginners, so let's slow down. Python does **not** use brackets to group code — it uses **indentation** (the spaces at the *start* of a line):

```text
if temperature < 5:
····print("Winter coat!")      ← indented 4 spaces → belongs to the if
····print("And a scarf.")      ← same indent → also belongs to the if
print("Have a nice day.")     ← back at the left edge → ALWAYS runs
```

Three rules that prevent 90% of beginner errors:

1. A line that opens a block **ends with a colon `:`** (`if ...:`, `else:`, later `for ...:` and `def ...:`).
2. Everything inside the block is indented by **4 spaces** (the notebook does this for you when you press Enter after a `:`).
3. The block ends when the indentation goes **back to the left**.

If you ever see `IndentationError` or `SyntaxError: expected ':'` — check exactly these three things.

````python
temperature = 12   # try changing this to 3, then to 25, and re-run!

if temperature < 5:
    print("🧥 Winter coat!")
elif temperature < 15:
    print("🧊 Light jacket weather.")
else:
    print("👕 T-shirt is fine.")

print(f"(Decision made for {temperature}°C — this line always runs.)")
````

**One branch, and only one.** Python checks the conditions from top to bottom and runs the **first** one that is `True` — the rest are skipped entirely. That's why the order matters: at 3°C, *both* `temperature < 5` and `temperature < 15` are true, but only the first branch fires.

## 3. `for` loops — doing something for *every* item

### The intuition: a conveyor belt

Imagine your list is a conveyor belt. A `for` loop picks up each item, **one at a time**, hands it to you under a name you choose, and runs your indented block once per item.

```text
for city in cities:        ← "for each item (call it `city`) in the list..."
····print(city)            ← ...run this block once per item
```

Same indentation rules as `if`: colon at the end, 4 spaces inside.

And when there is no list yet — you simply want to repeat something *n* times — use **`range(n)`**: it generates the numbers `0, 1, 2, …, n-1` for you (computers love starting at 0!).

````python
cities = ["Berlin", "Nairobi", "Osaka", "Lima"]

for city in cities:
    print(f"Hello, {city}!")

# range(n): repeat n times, counting 0 ... n-1
print()
for i in range(5):
    print(f"Round {i}")

# range(start, stop) — starts at `start`, stops JUST BEFORE `stop`
print()
for year in range(2024, 2028):
    print(f"Year {year}")
````

### The accumulator pattern — building up a result step by step

This is *the* most important loop pattern in data science. To sum a list by hand you would keep a **running total** on a piece of paper and add each number to it. Code does exactly the same:

1. **Before** the loop: create a variable that starts "empty" (e.g. `total = 0`).
2. **Inside** the loop: update it with each item.
3. **After** the loop: the result is ready.

````python
daily_steps = [8500, 12300, 6200, 9800, 15000]

total = 0                    # 1. start empty, BEFORE the loop
for steps in daily_steps:
    total = total + steps    # 2. update once per item

print(f"Total steps this week : {total}")            # 3. use the result
print(f"Average per day       : {total / len(daily_steps):.0f}")
````

## 4. Combining loop + `if` — filtering and counting

Now the two ideas click together: loop over your data, and *inside* the loop, use an `if` to react only to the items you care about. This "scan and count" pattern is the hand-made version of what every data-analysis tool does under the hood.

Note the **double indentation**: the `if` is indented once (inside the loop), and its body twice (inside the `if` inside the loop).

````python
temperatures = [19, 24, 31, 22, 28, 17, 33]   # one week, in °C

hot_days = 0
for temp in temperatures:
    if temp > 25:            # only react to hot days
        hot_days += 1        # += is shorthand for: hot_days = hot_days + 1

print(f"Hot days (>25°C) this week: {hot_days} out of {len(temperatures)}")
````

## 5. Your first function — giving a recipe a name

### The intuition

So far, if you wanted to repeat a computation with different inputs, you had to copy-paste code and edit it. A **function** fixes that: you write the recipe *once*, give it a **name**, and from then on you just say the name.

Think of a kitchen recipe card:

- **`def`** writes the card: *"To make `greet`, take one `name`, and do the following…"*
- **Parameters** are the blanks on the card ("take one `name`").
- **`return`** is the finished dish handed back to whoever asked.
- **Calling** the function (`greet("Ada")`) means: *"make me one, with this ingredient."*

```text
 define once                    call many times
┌───────────────────────┐       greet("Ada")   → "Hello, Ada!"
│ def greet(name):      │       greet("Bo")    → "Hello, Bo!"
│     return f"Hello..."│       greet("Chen")  → "Hello, Chen!"
└───────────────────────┘
```

And yes — the body of a function is indented, exactly like `if` and `for`. By now you know the drill: **colon, then 4 spaces.**

````python
def greet(name):
    return f"Hello, {name}! Welcome to Python."

# Defining it produced no output — we only wrote the recipe card.
# Now let's actually CALL it:
message = greet("Ada")
print(message)

print(greet("Bo"))       # call it as often as you like
print(greet("Chen"))


# Functions can take TWO parameters — and do real work:
def price_with_tip(price, tip_percent):
    tip = price * tip_percent / 100
    return price + tip

print(f"Bill: {price_with_tip(40, 10):.2f} €")   # 40 € plus 10% tip
print(f"Bill: {price_with_tip(25, 15):.2f} €")   # 25 € plus 15% tip

# `return` hands the value back, so you can store it and keep computing:
total = price_with_tip(40, 10) + price_with_tip(25, 15)
print(f"Both bills together: {total:.2f} €")
````

> 💡 **`return` vs `print`** — a subtle but important difference: `print` only *shows* a value on screen; `return` *hands it back* to your code so you can store it in a variable, add it to something, or feed it into another function. Recipes that `return` are reusable building blocks.

## 6. Putting it all together — a mini weather report 🌤️

Time for the payoff. We'll combine **everything** from this notebook in ~15 lines:

- a **function** that classifies a temperature into a category (`if / elif / else`),
- a **`for` loop** that applies it to a week of data,
- an **accumulator** counting the hot days,
- **f-strings** to print a tidy report.

Read the code first and predict what it will print — *then* run it.

````python
def classify(temp):
    """Turn a temperature in °C into a human-friendly category."""
    if temp < 10:
        return "cold 🧊"
    elif temp <= 25:
        return "pleasant 🙂"
    else:
        return "hot 🔥"


days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temps = [8, 19, 24, 31, 22, 28, 33]

print("=== Weekly Weather Report ===")
hot_days = 0
for i in range(len(days)):                 # i = 0, 1, 2, ... 6
    category = classify(temps[i])
    print(f"  {days[i]}: {temps[i]:>2}°C — {category}")
    if temps[i] > 25:
        hot_days += 1

print("=============================")
print(f"Hot days this week: {hot_days} of {len(days)}")
````

**Take a second to appreciate this.** One small function + one loop + one `if` — and you have a program that would work just as well on 7 days or 7 *million* days of data. That scalability is *the* reason we automate analysis with code instead of reading spreadsheets by eye. In Notebook 3, we'll apply this exact mindset to a real mini data analysis with a chart.

## 🧠 Key takeaways

1. **Booleans** (`True`/`False`) answer yes/no questions built with `==`, `!=`, `<`, `>` — and `and` / `or` / `not` combine them. Remember: `=` assigns, `==` compares.
2. **`if / elif / else`** is a decision tree: Python runs the **first true branch** and skips the rest — so order your conditions from strictest to loosest.
3. **Indentation is the grammar of Python**: a colon `:` opens a block, 4 spaces mark what's inside, and going back to the left edge closes it.
4. **`for` loops** visit every item of a list once; **`range(n)`** counts `0 … n-1` when there's no list.
5. The **accumulator pattern** (initialise before the loop → update inside → use after) builds sums and counts — combine it with `if` to filter.
6. **Functions** (`def` + parameters + `return`) give a recipe a name so you can reuse it — `return` hands the value back, `print` only displays it.
