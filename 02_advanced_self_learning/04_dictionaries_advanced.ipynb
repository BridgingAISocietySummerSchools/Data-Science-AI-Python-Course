## 1. What is a dictionary?

Think of a dictionary as a **labelled filing cabinet**: each drawer has a **label** (the *key*) and contains some *value*.

```
       ┌───── student ─────┐
"name"   →  "Alice"
"age"    →  20
"major"  →  "Computer Science"
"gpa"    →  3.8
```

- Keys must be **unique** and **immutable** (strings, numbers, tuples — *not* lists).
- Values can be anything (numbers, strings, lists, even other dictionaries).
- Dictionaries are **ordered** since Python 3.7 (keys keep their insertion order).

````python
# Creating a dictionary
student = {
    "name":  "Alice",
    "age":   20,
    "major": "Computer Science",
    "gpa":   3.8,
}

print(student)
print(f"type = {type(student).__name__}, size = {len(student)}")

# Empty dict
inventory = {}
print(f"empty: {inventory}")
````

## 2. Accessing values

Use `dict[key]` to look up a value. Like list-indexing, but with a meaningful label instead of a position.

````python
student = {"name": "Alice", "age": 20, "major": "CS", "gpa": 3.8}

print(f"Name : {student['name']}")
print(f"Age  : {student['age']}")
print(f"GPA  : {student['gpa']}")

# What if the key doesn't exist?
try:
    print(student["height"])
except KeyError as e:
    print(f"\nKeyError: missing key {e}")

# Safer: .get() returns None (or a default) when the key is missing
print(f"\n.get('height') -> {student.get('height')}")
print(f".get('height', 'unknown') -> {student.get('height', 'unknown')}")

# Or simply ask first — `in` checks the KEYS of a dictionary
print(f"\n'gpa' in student    -> {'gpa' in student}")
print(f"'height' in student -> {'height' in student}")
````

> 💡 **Rule of thumb:** use `d[key]` when a missing key would be a **bug you want to hear about** — the loud `KeyError` points you straight at it. Use `d.get(key, default)` when missing keys are **normal** (optional fields, messy real-world data). And `key in d` lets you ask before you touch.

## 3. Adding, updating and removing entries

````python
person = {"name": "Bob", "age": 25}
print(f"Original : {person}")

# Add new keys
person["city"] = "Berlin"
person["job"]  = "Data Analyst"
print(f"After add: {person}")

# Update an existing key (same syntax — there's no difference)
person["age"] = 26
print(f"After update: {person}")

# Update multiple keys at once
person.update({"city": "Munich", "salary": 65_000})
print(f"After bulk update: {person}")

# Remove keys
removed = person.pop("salary")   # remove AND return the value
print(f"Popped 'salary' = {removed}")
print(f"After pop: {person}")

del person["city"]
print(f"After del 'city': {person}")
````

## 4. Useful methods: `keys()`, `values()`, `items()`

````python
inventory = {"apples": 50, "bananas": 30, "oranges": 25, "grapes": 40}

print(f"Keys   : {list(inventory.keys())}")
print(f"Values : {list(inventory.values())}")
print(f"Items  : {list(inventory.items())}")

# Numbers you can compute directly
print(f"\nTotal stock : {sum(inventory.values())}")
print(f"Most stocked: {max(inventory, key=inventory.get)}")   # neat trick
````

> 🎯 **`max(d, key=d.get)`** finds the key whose value is largest — a one-liner version of "which item has the most stock?".

## 5. Iterating over a dictionary

Three flavours, depending on what you need:

````python
grades = {"Math": 85, "Science": 92, "English": 78, "History": 88}

# 1) Iterating directly gives KEYS
print("Keys only:")
for subject in grades:
    print(f"  {subject}")

# 2) Just the values
print("\nValues only:")
for score in grades.values():
    print(f"  {score}")

# 3) Both at once — the most common pattern
print("\nKey + value (using .items()):")
for subject, score in grades.items():
    print(f"  {subject:<10} {score}")
````

## 6. Dictionary comprehensions

The same `{expression for ... in ... if ...}` syntax as list comprehensions, but produces a dictionary.

````python
# Build a dict from scratch
squares = {x: x**2 for x in range(1, 6)}
print("Squares:", squares)

# Transform values
celsius     = {"morning": 20, "noon": 30, "evening": 25}
fahrenheit  = {k: (v * 9/5 + 32) for k, v in celsius.items()}
print(f"\nC: {celsius}")
print(f"F: {fahrenheit}")

# Filter — keep only items meeting a condition
scores  = {"Alice": 85, "Bob": 92, "Charlie": 67, "Diana": 78, "Eve": 95}
high    = {name: s for name, s in scores.items() if s >= 80}
print(f"\nHigh scorers: {high}")

# Invert — swap keys and values (works when values are unique)
inverted = {v: k for k, v in scores.items()}
print(f"\nInverted: {inverted}")
````

## 7. Nested dictionaries — real-world structured data

In practice, a dictionary value is often *another* dictionary. This is how you'd represent records from a database, JSON from an API or settings from a config file.

````python
students_db = {
    "S001": {
        "name":   "Alice Johnson",
        "age":    20,
        "major":  "Computer Science",
        "grades": {"Math": 85, "Science": 92, "English": 78},
    },
    "S002": {
        "name":   "Bob Smith",
        "age":    22,
        "major":  "Mathematics",
        "grades": {"Math": 95, "Science": 88, "English": 82},
    },
    "S003": {
        "name":   "Carol Lee",
        "age":    21,
        "major":  "Physics",
        "grades": {"Math": 90, "Science": 95, "English": 75},
    },
}

# Drill down with chained keys
print(students_db["S001"]["grades"]["Math"])     # → 85

# Walk the database and compute each student's average grade
for sid, info in students_db.items():
    g = info["grades"]
    avg = sum(g.values()) / len(g)
    print(f"  [{sid}] {info['name']:<14} avg = {avg:5.2f}")
````

## 8. List of dictionaries — the shape of (almost) all data

Here comes the single most important idea in this notebook. When you fetch data from a web API or read a CSV file, what you get is almost always a **list of dictionaries**: one dict per **row**, with the column names as keys.

```
employees = [
    {"name": "Alice", "department": "Engineering", "salary": 75000, "years": 3},   # ← row 1
    {"name": "Bob",   "department": "Marketing",   "salary": 65000, "years": 5},   # ← row 2
    ...
]
```

Three names, one shape:

| You call it…        | It looks like…                            |
|---------------------|-------------------------------------------|
| a **table**         | rows × columns, like a spreadsheet        |
| a **list of dicts** | `[{...}, {...}, ...]` in Python           |
| **JSON**            | what practically every web API sends you  |

Learn to manipulate this one shape and you can handle real data *today* — and in Notebook 7 ([`07_pandas_essentials.ipynb`](07_pandas_essentials.ipynb)) you'll see that `pd.DataFrame(employees)` turns it into a pandas table in a single call.

````python
employees = [
    {"name": "Alice",   "department": "Engineering", "salary": 75000, "years": 3},
    {"name": "Bob",     "department": "Marketing",   "salary": 65000, "years": 5},
    {"name": "Charlie", "department": "Engineering", "salary": 80000, "years": 4},
    {"name": "Diana",   "department": "Sales",       "salary": 60000, "years": 2},
    {"name": "Eve",     "department": "Engineering", "salary": 90000, "years": 7},
    {"name": "Frank",   "department": "Marketing",   "salary": 70000, "years": 6},
]

# Average salary
avg_salary = sum(e["salary"] for e in employees) / len(employees)
print(f"Average salary : €{avg_salary:,.0f}")

# Highest-paid employee
top = max(employees, key=lambda e: e["salary"])
print(f"Top earner     : {top['name']} (€{top['salary']:,})")

# Sort by years of experience, descending
ranked = sorted(employees, key=lambda e: e["years"], reverse=True)
print("\nRanked by experience:")
for r in ranked:
    print(f"  {r['name']:<8} {r['years']:>2}y in {r['department']}")
````

> 💡 **What is `lambda e: e["salary"]`?** A *lambda* is a tiny unnamed function written inline: "given a row `e`, hand back `e['salary']`". Passing it as `key=` tells `max` / `sorted` **which field to compare on** — Python calls it once per row and ranks the rows by what comes back. We'll study lambdas properly in Notebook 5 ([`05_functions_and_modules.ipynb`](05_functions_and_modules.ipynb)); until then, `key=lambda row: row["column"]` is a recipe worth memorising.

## 9. Counting and grouping — dictionaries' superpower

Counting categories and grouping rows by a column are textbook dictionary jobs. The pattern below appears in **every** data-cleaning project.

````python
# Count survey responses
responses = ["Yes", "No", "Yes", "Maybe", "Yes", "No", "Yes", "Maybe", "Yes", "No"]

counts = {}
for r in responses:
    counts[r] = counts.get(r, 0) + 1     # idiom: get-or-default 0, then add 1

print(f"Counts: {counts}")
````

````python
# Even cleaner with collections.Counter (built into Python)
from collections import Counter

responses = ["Yes", "No", "Yes", "Maybe", "Yes", "No", "Yes", "Maybe", "Yes", "No"]
print(Counter(responses))                # like a dict, with a few extras
print(Counter(responses).most_common(2)) # top-2 most common values
````

````python
# Group employees by department — list of dicts → dict of lists
employees = [
    {"name": "Alice",   "department": "Engineering", "salary": 75000},
    {"name": "Bob",     "department": "Marketing",   "salary": 65000},
    {"name": "Charlie", "department": "Engineering", "salary": 80000},
    {"name": "Diana",   "department": "Sales",       "salary": 60000},
    {"name": "Eve",     "department": "Engineering", "salary": 90000},
    {"name": "Frank",   "department": "Marketing",   "salary": 70000},
]

# The manual pattern — same spirit as counting, but collecting into lists
by_dept = {}
for e in employees:
    dept = e["department"]
    if dept not in by_dept:      # first time we see this department?
        by_dept[dept] = []       # ...start an empty list for it
    by_dept[dept].append(e["name"])

print("Employees by department:")
for dept, names in by_dept.items():
    print(f"  {dept:<12}: {names}")

# defaultdict removes the "is this key new?" dance: missing keys start as an empty list
from collections import defaultdict

salaries_by_dept = defaultdict(list)
for e in employees:
    salaries_by_dept[e["department"]].append(e["salary"])

print("\nAverage salary by department:")
for dept, salaries in salaries_by_dept.items():
    print(f"  {dept:<12}: €{sum(salaries)/len(salaries):,.0f}")
````

> 🔭 **Where this is heading.** Everything in this section is *one line* in pandas:
>
> ```python
> pd.DataFrame(employees).groupby("department")["salary"].mean()
> ```
>

## 10. Lists vs dictionaries — when to use which?

| Use a **list** when…                                       | Use a **dict** when…                                      |
|------------------------------------------------------------|-----------------------------------------------------------|
| Order matters (time series, ranks, instructions).          | Order is incidental; lookup-by-name matters.              |
| You'll access items by **position** (`x[0]`, `x[1]`).      | You'll access items by **label** (`x["name"]`).            |
| Duplicates are fine and meaningful.                        | Keys must be unique.                                       |
| Items are *similar* and interchangeable.                   | Items are *named* with different meanings.                 |

> 🎯 In data science, *records* (rows) are usually dicts; *columns* are usually lists/arrays.

## 11. A tiny taste of JSON

Dictionaries map directly to **JSON** — the universal data-exchange format of the web. `json.dumps` turns a Python object into a JSON *string* (ready to send or save); `json.loads` turns it back. Notice below that the JSON text is almost character-for-character the Python you've been writing all notebook.

````python
import json

# A single record is a dict...
api_response = {
    "city": "Berlin",
    "temperature_c": 21.4,
    "conditions": ["cloudy", "windy"],
    "forecast": {"tomorrow": "rain", "weekend": "sunny"},
}
print(json.dumps(api_response, indent=2))

# Round-trip: dict → JSON string → dict
s = json.dumps(api_response)
back = json.loads(s)
print(f"\nRound-trip equal? {back == api_response}")

# ...and a TABLE is a list of dicts — this is literally what most APIs send you
team = [
    {"name": "Alice", "department": "Engineering", "salary": 75000},
    {"name": "Bob",   "department": "Marketing",   "salary": 65000},
]
print("\n" + json.dumps(team, indent=2))
````

### Exercise 4.4 — Sensor data pipeline

Real sensors often deliver anonymous tuples like the ones below — and position-based code (`r[1]`, `r[2]`) is exactly the hard-to-read style this notebook teaches you to escape. So:

```python
# 1. Tuples → records: give every field a NAME
readings = [
    {"time": t, "temp": temp, "humidity": hum}
    for t, temp, hum in sensor_readings
]

# 2. Aggregates — notice how readable the field names are vs. r[1] / r[2]
temps = [r["temp"] for r in readings]
hums  = [r["humidity"] for r in readings]
print(f"Mean temp     : {sum(temps)/len(temps):.2f} °C")
print(f"Mean humidity : {sum(hums)/len(hums):.1f} %")

print(f"Peak humidity : {peak['humidity']}% at {peak['time']}")

# 3. Hot readings
hot = [r for r in readings if r["temp"] >= 24]

    print(f"  {r['time']}  {r['temp']} °C")
```

**Reasoning.** Step 1 is the pay-off move: unpacking each tuple (`for t, temp, hum in ...`) inside a list comprehension turns anonymous positions into named fields, and every later step then reads like plain English. ## 🧠 Key takeaways

1. Dictionaries store **key → value** pairs; keys must be unique and immutable.
2. Use `d[k]` to read **when you're sure the key exists**, otherwise `d.get(k, default)`.
3. `.keys()`, `.values()`, `.items()` are how you iterate; **`.items()` is by far the most useful**.
4. Use a **list of dictionaries** for tabular data (one dict per row).
5. The `dict.get(k, 0) + value` idiom is the classic "count / sum by group" pattern.
6. `Counter` and `defaultdict` from `collections` make counting and grouping painless.
7. Dictionaries map directly to **JSON** — you'll see this shape in every web API.
