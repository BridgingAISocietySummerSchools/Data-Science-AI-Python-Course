## 2. Variables — labelled boxes for your data

Before a computer can *work* with a value, it needs somewhere to *keep* it. A **variable** is a name attached to a value — think of it as a **labelled box**:

```
┌──────────┐    ┌───────┐    ┌──────────┐
│  "Alice" │    │  25   │    │   1.70   │
└────┬─────┘    └───┬───┘    └────┬─────┘
     │              │             │
   name            age         height
```

You put a value into a box with the `=` sign (read it as *"gets"*, not *"equals"*): `age = 25` means *"the box labelled `age` gets the value 25"*. From then on, writing `age` anywhere means *"look inside that box"*.

````python
name = "Alice"      # text goes in quotes
age = 25            # a whole number
height = 1.70       # a decimal number

print(name)
print(age)
print(height)

# Variables can be reused and combined:
birth_year = 2026 - age
print(birth_year)
````

**Reading the output:** the first three lines simply show what is inside each box. The last line shows the real power — we *computed* a new value (`birth_year`) from an existing one. If `age` changes, we only edit one place and everything downstream updates.

Two naming rules to remember (there are more, but these carry you far):

- Names are written in `snake_case`: lowercase words joined by underscores, e.g. `birth_year`, `coffee_price`.
- Choose names that say what the value *means* — `temperature` beats `t`.

## 3. Data types — not all values are alike

Look at a spreadsheet from any field: some columns hold whole numbers (participants), some hold decimals (measurements), some hold text (city names), and some hold yes/no answers. Python makes the same distinction. The four types you will meet every single day:

| Type | Python name | Example | Typical use |
|------|-------------|---------|-------------|
| Whole number | `int` | `42` | counts, years, IDs |
| Decimal number | `float` | `19.99` | measurements, prices, averages |
| Text ("string") | `str` | `"Berlin"` | names, labels, answers |
| Yes/No | `bool` | `True`, `False` | flags, comparison results |

You can always *ask* Python what type a value has, using `type()`:

````python
participants = 42          # int
average_temp = 21.5        # float
city = "Berlin"            # str
survey_complete = True     # bool

print(type(participants))
print(type(average_temp))
print(type(city))
print(type(survey_complete))
````

**Why care?** Because types decide what you can *do* with a value. You can multiply two numbers, but `"Berlin" * "Hamburg"` makes no sense — and Python will say so. The most common beginner trip-up: `"42"` (with quotes) is *text*, not a number. `"42" + 1` produces an error; `42 + 1` produces `43`.

## 4. Arithmetic — Python as a very reliable calculator

Python knows all the operators from school maths, plus a couple of handy extras:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | addition | `7 + 3` | `10` |
| `-` | subtraction | `7 - 3` | `4` |
| `*` | multiplication | `7 * 3` | `21` |
| `/` | division | `7 / 2` | `3.5` |
| `**` | power | `2 ** 10` | `1024` |
| `//` | division, rounded down | `7 // 2` | `3` |
| `%` | remainder of a division | `7 % 2` | `1` |

The last two look exotic but are surprisingly useful: `//` answers *"how many whole times does it fit?"* and `%` answers *"what is left over?"* — think of splitting 7 pizza slices among 2 people.

Let's use variables and arithmetic for a tiny real calculation — what your daily coffee habit costs per year:

````python
coffee_price = 3.20        # euros per cup
cups_per_day = 2
days_per_year = 365

yearly_cost = coffee_price * cups_per_day * days_per_year
print("Coffee per year, in euros:")
print(yearly_cost)

# The extras in action: splitting 7 slices among 2 people
print(7 // 2)   # each person gets 3 whole slices ...
print(7 % 2)    # ... and 1 slice is left over
````

## 5. Strings and f-strings — talking about your results

A number on its own (`16.5`) is not a result — *"Each person pays €16.50"* is. Combining text and values is such a common job that Python has a dedicated tool for it: the **f-string**.

Put an `f` before the opening quote, and you can embed any variable inside curly braces `{ }`. Even better: after a colon you can add a **format instruction**. The two you will use most in data work:

| Spec | Meaning | `f"{1234.5678:...}"` gives |
|------|---------|-----------------------------|
| `:.2f` | exactly 2 decimal places | `1234.57` |
| `:,` | thousands separator | `1,234.5678` |
| `:,.2f` | both combined | `1,234.57` |

````python
name = "Alice"
age = 25
per_person = 16.5
big_number = 1234567.891

# Embedding values — even small calculations work inside { }
print(f"{name} is {age} years old.")
print(f"Next year, {name} will be {age + 1}.")

# Formatting numbers
print(f"Each person pays €{per_person:.2f}")           # 2 decimals: 16.50
print(f"A big number, readable: {big_number:,.2f}")    # separators + 2 decimals
````

## 6. Lists — many values in one box

So far every box held **one** value. Real data comes in *series*: temperatures for a week, answers from 200 survey participants, prices over a year. A **list** is a box with numbered compartments — you create one with square brackets:

```
temps  =  [ 18.5 , 21.0 , 19.2 , 23.1 , 20.7 ]
position:    0      1      2      3      4
```

One crucial convention: **Python starts counting at 0.** The first item is item `0`, the second is item `1`, and so on.

````python
temps = [18.5, 21.0, 19.2, 23.1, 20.7]   # daily temperatures, Mon-Fri

print(temps)          # the whole list
print(temps[0])       # first item  -> 18.5   (counting starts at 0!)
print(temps[3])       # fourth item -> 23.1
print(temps[-1])      # negative index counts from the END: last item -> 20.7
print(len(temps))     # len() tells you how many items the list holds -> 5
````

````python
# Lists can also GROW. The weekend arrives, we measure again,
# and .append() adds the new value to the end:
temps.append(24.3)

print(temps)          # the list now ends with 24.3
print(len(temps))     # now 6 items
````

**Reading the output:** in the first cell, `temps[0]` is Monday and `temps[3]` is Thursday — remember, counting starts at 0. The negative index `temps[-1]` is a lovely shortcut for *"the last one"* — no need to know how long the list is. And after `.append()`, the list itself has changed: it is one item longer.

## 7. Putting it all together — your first mini analysis

Time for the payoff. Python has a few built-in helpers that work directly on lists of numbers:

- `sum(temps)` — add everything up
- `len(temps)` — how many values (so `sum(...) / len(...)` is the **average**)
- `min(temps)` / `max(temps)` — smallest and largest value

With variables, arithmetic, f-strings and a list, we can write a small but *genuine* data analysis — the same logic behind every "average temperature" you have ever seen in a weather report:

````python
# One week of daily temperatures (°C)
temps = [18.5, 21.0, 19.2, 23.1, 20.7, 24.3, 22.8]

# The analysis: three lines of real statistics
average = sum(temps) / len(temps)
coldest = min(temps)
warmest = max(temps)

# The report: f-strings make it readable
print("🌡️  WEEKLY TEMPERATURE REPORT")
print("-" * 34)
print(f"Days measured : {len(temps)}")
print(f"Average       : {average:.2f} °C")
print(f"Coldest day   : {coldest:.1f} °C")
print(f"Warmest day   : {warmest:.1f} °C")
print(f"Range         : {warmest - coldest:.1f} °C")
````

**What just happened?** Everything from this notebook, working together:

1. A **list** stored the raw data in one place.
2. **Arithmetic** with `sum()` and `len()` derived the average — no calculator, no copy-paste.
3. **f-strings** with `:.2f` turned raw numbers into a report a human wants to read.

Change one temperature in the list and re-run the cell: the entire report updates itself. That, in miniature, is what data analysis with Python *is* — and in Notebook 3 you will do it on real data, with a chart.
