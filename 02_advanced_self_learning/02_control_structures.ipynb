## 1. Python's indentation system — the "shape" of code

Most other languages use braces `{ … }` to group code. Python uses **indentation**: lines that belong to the same block must start at the same column.

```
if temperature > 20:
····print("It's warm outside!")     ← 4 spaces in → part of the if-block
····print("Enjoy your day.")        ← also part of it
print("This always prints.")       ← back to column 0 → outside the if
```

Rules of thumb:

- The standard is **4 spaces** per level. Don't mix spaces and tabs.
- A "block-opening" line ends in `:` (e.g. `if x > 0:`, `for i in range(3):`, `def f():`).
- The block ends when indentation goes back to the previous level.

Most editors (including Colab) insert 4 spaces automatically when you press Tab after a `:` line. Try the example below — change `temperature` and re-run.

````python
temperature = 25

if temperature > 20:
    print("It's warm outside ☀️")
    print(f"  Reading: {temperature}°C")
print("After the if-block (always runs).")
````

> ⚠️ **IndentationError** is the #1 error new Python users see. If you see it, look at the column where each line *starts*. Often the issue is mixing tabs and spaces.

## 2. `if / elif / else` — making decisions

The structure mirrors how we describe rules in natural language:

> *"If the score is at least 90, give an A; otherwise if it's at least 80, give a B; otherwise..."*

````python
# Two-way decision: if / else
score = 75

if score >= 70:
    print("✅ You passed!")
else:
    print("❌ You did not pass. Try again.")
````

````python
# Multi-way decision: if / elif / else
def grade_for(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

# Quick check across several scores
for s in [95, 82, 71, 64, 41]:
    print(f"  {s} → grade {grade_for(s)}")
````

**How Python evaluates `elif`.** Python checks the branches *in order* and runs the **first** one whose condition is true. The rest are skipped. So order matters — if you put `score >= 60` before `score >= 90`, *every* passing score gets a D.

> 🎯 **Intuition.** Think of `if/elif/else` as a fall-through staircase. As soon as a step is "true", you take it; you never check lower steps.

### Comparison and logical operators — your decision toolkit

| Operator | Meaning            | Example          | Result   |
|----------|--------------------|------------------|----------|
| `==`     | equal              | `5 == 5`         | `True`   |
| `!=`     | not equal          | `5 != 3`         | `True`   |
| `<`      | less than          | `3 < 5`          | `True`   |
| `>`      | greater than       | `3 > 5`          | `False`  |
| `<=`     | less or equal      | `5 <= 5`         | `True`   |
| `>=`     | greater or equal   | `5 >= 6`         | `False`  |
| `and`    | logical AND        | `True and False` | `False`  |
| `or`     | logical OR         | `True or False`  | `True`   |
| `not`    | logical NOT        | `not True`       | `False`  |
| `in`     | contains           | `"a" in "cat"`   | `True`   |

> 💡 **A common bug.** People write `=` when they mean `==`. `=` *assigns*; `==` *compares*. `if x = 5:` is a syntax error in Python (which is a feature, not a bug — it stops the mistake).

````python
age          = 25
has_license  = True
has_car      = False

# Combine conditions with `and` / `or`
if age >= 18 and has_license:
    print("Legally allowed to drive ✅")

if has_car or age >= 25:
    print("Eligible for the rental discount 💸")

# Negate with `not`
if not has_car:
    print("Note: no car owned yet.")

# Chained comparison — Pythonic shortcut for "x between a and b"
working_age = 18 <= age <= 65
print(f"In working age range? {working_age}")
````

### Truthiness — when "falsy" values trip you up

Python treats several values as **falsy** in a boolean context: `0`, `0.0`, `""` (empty string), `None`, and empty containers like `[]`, `{}`, `()`. Everything else is **truthy**.

This is convenient (`if my_list:` reads as "if the list is non-empty"), but watch out — `if temperature:` is True even when temperature is `-30`, but **False** when temperature is exactly `0`.

````python
for value in [0, 1, -1, 0.0, "", "hi", None, [], [0], {}]:
    print(f"{repr(value):>6} → {'truthy' if value else 'falsy'}")
````

## 3. `for` loops — repeating with a counter or over data

A `for` loop iterates over a *sequence*. Two common patterns:

1. **`range(...)`** — when you need to repeat *n* times or count.
2. **Over a list / tuple / string** — when you have actual data.

````python
# Pattern 1: range(stop)  — counts 0, 1, 2, ..., stop-1
print("range(5):")
for i in range(5):
    print(f"  i = {i}")

print("\nrange(1, 6):  (start inclusive, stop exclusive)")
for i in range(1, 6):
    print(f"  i = {i}")

print("\nrange(0, 11, 2):  (step)")
for i in range(0, 11, 2):
    print(f"  i = {i}")
````

````python
# Pattern 2: iterate over a real collection
cities = ["Berlin", "Munich", "Hamburg", "Cologne"]
for city in cities:
    print(f"  {city}: population unknown")
````

### `enumerate` — when you need both the index *and* the value

````python
scores = [85, 92, 78, 96, 88]

for index, value in enumerate(scores):
    print(f"Score #{index}: {value}")

# Start counting at 1 if you prefer 1-based labels
print("\n1-based version:")
for rank, value in enumerate(scores, start=1):
    print(f"  Rank {rank}: {value}")
````

### Accumulating in a loop — the classic data-science pattern

Whenever you compute a sum, count, min, max or mean *by hand*, the recipe is always: start with an initial value, then update it inside the loop.

````python
temperatures = [18.5, 21.0, 19.7, 22.3, 25.1, 23.8, 20.4]

# Pattern: initialise outside the loop, update inside
total = 0
n     = 0

for temp in temperatures:
    total += temp
    n     += 1

mean = total / n
print(f"Days     : {n}")
print(f"Sum      : {total:.1f}")
print(f"Mean temp: {mean:.2f}°C")

# Python has a built-in for this:  sum() / len()
print(f"\nDouble-check with built-ins: {sum(temperatures)/len(temperatures):.2f}°C")
````

> 🎯 **Mental model.** A `for` loop on a sequence is like a conveyor belt: each item walks past once, you do something with it, then it disappears.

## 4. `while` loops — repeat while a condition is true

Use `while` when you don't know in advance *how many* iterations are needed — only when to stop.

````python
# Compound-interest example: how many years until the savings double?
balance = 1000
target  = 2000
rate    = 0.05
years   = 0

while balance < target:
    balance *= (1 + rate)
    years   += 1
    print(f"  Year {years}: €{balance:,.2f}")

print(f"\n→ Doubles after {years} years at {rate:.0%} annual interest.")
````

> ⚠️ **Infinite-loop trap.** A `while` loop *only* stops when the condition becomes false. Make sure the variable in the condition is actually updated inside the loop, otherwise the cell will run forever. In Colab you can stop it with the ⏹️ button (or **Runtime → Interrupt execution**).

## 5. `break` and `continue` — fine-grained control

- **`break`** stops the loop entirely (exit early).
- **`continue`** skips the rest of *this* iteration and starts the next.

These let you avoid deeply nested `if`s and keep code readable.

````python
# break: stop searching once you find a match
print("Looking for the first multiple of 7:")
for n in range(1, 100):
    if n % 7 == 0:
        print(f"  Found it: {n}")
        break
````

````python
# continue: skip the negatives ("dirty" data points)
raw = [12, -3, 8, -1, 15, 0, 22, -7]

clean_total = 0
n_valid = 0
for x in raw:
    if x <= 0:
        continue                       # skip invalid measurements
    clean_total += x
    n_valid     += 1

print(f"Valid points : {n_valid}")
print(f"Clean mean   : {clean_total/n_valid:.2f}")
````

## 6. Nested loops — for table-shaped data

A loop *inside* another loop is useful for two-dimensional structures.

````python
# Mini multiplication table
size = 4
for i in range(1, size + 1):
    row = ""
    for j in range(1, size + 1):
        row += f"{i*j:4d}"             # 4 char width
    print(row)
````

## 7. Handling errors gracefully — `try / except`

In real data work some rows will be missing, some values will be invalid, some divisions will hit zero. Instead of crashing the whole notebook, you can **catch** the error and continue:

````python
readings = ["12.4", "missing", "0", "15.8", "n/a", "20.1"]

valid_values = []
for raw in readings:
    try:
        value = float(raw)               # may fail with ValueError
        result = 100 / value             # may fail with ZeroDivisionError
        valid_values.append(value)
        print(f"  OK   {raw!r:>10}  →  {value:6.2f}   (100/value = {result:.2f})")
    except ValueError:
        print(f"  SKIP {raw!r:>10}  →  not a number")
    except ZeroDivisionError:
        print(f"  SKIP {raw!r:>10}  →  divides by zero")

print(f"\nValid points kept: {len(valid_values)} / {len(readings)}")
````

> 💡 Catch **specific** exceptions, not a bare `except:` — that hides bugs you would actually want to know about.

## 8. Putting it all together — a tiny data-analysis pipeline

Let's combine *everything* from this notebook into a small but realistic task: given a list of raw daily temperatures (with occasional bad values), compute summary statistics and categorise each day.

````python
# Raw temperatures with some bad data
raw_temps = ["18.5", "21.0", "n/a", "19.7", "22.3", "", "25.1",
             "23.8", "missing", "20.4", "-273.15", "27.0"]

valid_temps = []
labels      = []

for raw in raw_temps:
    try:
        t = float(raw)
    except ValueError:
        continue                  # discard non-numeric

    # Filter physically impossible values (below absolute zero in °C ≈ -273.15)
    if t < -50 or t > 60:
        continue

    valid_temps.append(t)

    # Categorise the day
    if   t < 5:   labels.append("very cold")
    elif t < 15:  labels.append("cold")
    elif t < 25:  labels.append("pleasant")
    else:         labels.append("hot")

# Summary statistics by hand (and with built-ins)
n = len(valid_temps)
mean = sum(valid_temps) / n
print(f"Kept {n}/{len(raw_temps)} valid readings")
print(f"  min  : {min(valid_temps):.1f}°C")
print(f"  mean : {mean:.2f}°C")
print(f"  max  : {max(valid_temps):.1f}°C")

# Count each category
print("\nCategory counts:")
for cat in ("very cold", "cold", "pleasant", "hot"):
    count = sum(1 for L in labels if L == cat)   # tiny generator + sum
    print(f"  {cat:<10}: {count}")
````

**What this little script demonstrates.**

- `try/except` for parsing dirty inputs
- `if/elif/else` for categorisation
- list `append` accumulation in a `for` loop
- `min`, `max`, `sum`, `len` — Python built-ins that work over any list of numbers
- a generator expression `sum(1 for L in labels if L == cat)` for counting — a compact way to count items matching a predicate

These patterns repeat over and over in data science. Pandas just wraps them in nicer syntax.

## 🧠 Key takeaways
